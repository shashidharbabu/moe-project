# Exploratory Data Analysis

This notebook is used for exploratory data analysis on the datasets used in the MoE-MID project. The goal is to understand the characteristics of the datasets, visualize the data distributions, and identify any potential issues before training the models.

```python
import os
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Set the style for seaborn
sns.set(style='whitegrid')

# Define the paths to the datasets
ffhq_path = '../data/raw/ffhq'
food101_path = '../data/raw/food101'
places365_path = '../data/raw/places365'

# Function to count images in a directory
def count_images(directory):
    return len([name for name in os.listdir(directory) if name.endswith(('.png', '.jpg', '.jpeg'))])

# Count images in each dataset
ffhq_count = count_images(ffhq_path)
food101_count = count_images(food101_path)
places365_count = count_images(places365_path)

# Create a DataFrame for visualization
data_counts = pd.DataFrame({
    'Dataset': ['FFHQ', 'Food-101', 'Places365'],
    'Image Count': [ffhq_count, food101_count, places365_count]
})

# Plot the image counts
plt.figure(figsize=(10, 6))
sns.barplot(x='Dataset', y='Image Count', data=data_counts, palette='viridis')
plt.title('Number of Images in Each Dataset')
plt.xlabel('Dataset')
plt.ylabel('Image Count')
plt.show()
```
