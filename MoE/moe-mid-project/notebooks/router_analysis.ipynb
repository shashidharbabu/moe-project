# Router Analysis

This notebook analyzes the performance of the router models trained in the MoE-MID project. It evaluates the effectiveness of the student router in predicting domain probabilities based on text prompts and compares its performance against the teacher router.

```python
import torch
from src.models.router_student import RouterStudent
from src.models.router_teacher import RouterTeacher
from src.utils.logging import setup_logging
from src.evaluation.clip_score import calculate_clip_score
from src.evaluation.fid_kid import calculate_fid, calculate_kid
from tools.prompt_sets.domain_prompts import load_prompts

setup_logging()

# Load the trained models
student_router = RouterStudent.load_from_checkpoint('path/to/student_router_checkpoint.pt')
teacher_router = RouterTeacher.load_from_checkpoint('path/to/teacher_router_checkpoint.pt')

# Load prompts for evaluation
prompts = load_prompts()

# Evaluate the routers
def evaluate_router(router, prompts):
    results = []
    for prompt in prompts:
        soft_labels = router.predict(prompt)
        results.append(soft_labels)
    return results

student_results = evaluate_router(student_router, prompts)
teacher_results = evaluate_router(teacher_router, prompts)

# Calculate evaluation metrics
clip_scores_student = calculate_clip_score(student_results)
clip_scores_teacher = calculate_clip_score(teacher_results)

fid_score_student = calculate_fid(student_results)
fid_score_teacher = calculate_fid(teacher_results)

kid_score_student = calculate_kid(student_results)
kid_score_teacher = calculate_kid(teacher_results)

# Display results
print("Student Router CLIP Score:", clip_scores_student)
print("Teacher Router CLIP Score:", clip_scores_teacher)
print("Student Router FID Score:", fid_score_student)
print("Teacher Router FID Score:", fid_score_teacher)
print("Student Router KID Score:", kid_score_student)
print("Teacher Router KID Score:", kid_score_teacher)
```

## Conclusion

This notebook provides a framework for analyzing the performance of the router models. The results from the evaluation metrics will help in understanding the effectiveness of the student router compared to the teacher router, guiding future improvements.
